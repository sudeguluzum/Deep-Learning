Veriler örnek cümleler üzerinden N Gram (unigram, biagram, trigram) Countvectorizer ile analizi

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
belgeler=["Bu çalışma bir NGRAM çalışmasıdır.",
          "Bu çalışma doğal dil işleme çalışmasıdır."]

#unigram
unigram_model=CountVectorizer(ngram_range=(1,1))

#bigram
bigram_model=CountVectorizer(ngram_range=(2,2))

#trigram
trigram_model=CountVectorizer(ngram_range=(3,3))


#unigram analizi
X_unigram= unigram_model.fit_transform(belgeler) #sayısal vektörlere dönüştürme
unigram_ozellikleri= unigram_model.get_feature_names_out() #kelime listesi

#bigram analizi
X_bigram= bigram_model.fit_transform(belgeler)
bigram_ozellikleri = bigram_model.get_feature_names_out()

#trigram analizi
X_trigram= trigram_model.fit_transform(belgeler)
trigram_ozellikleri = trigram_model.get_feature_names_out()

print(f"Unigram: {unigram_ozellikleri}")
print(f"Bigram: {bigram_ozellikleri}")
print(f"Trigram: {trigram_ozellikleri}")
```

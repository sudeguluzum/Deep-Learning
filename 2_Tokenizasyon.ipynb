Metni daha küçük parçalara ayırma işlemidir

-kelime tokenizasyonu

-cümle tokenizasyonu



```python
import nltk # natural language took kit
#nltk.download("punkt") #kelime ve cümle tokenizasyonu için gerekli veri
nltk.download("punkt_tab")
```

```python
#örnek metin tanımla
raw_text= "Merhaba Google! Bu bir NLP eğitimidir. NLP eğitiminin ilerleyen aşamalarında LLM konusunu öğrenelim."

#kelime tokenizasyonu
word_tokens = nltk.word_tokenize(raw_text)
print(f"Kelime tokens: {word_tokens}")


#cümle tokenizasyonu
sentence_tokens=nltk.sent_tokenize(raw_text)
print(f"Cümle tokens: {sentence_tokens}")
```

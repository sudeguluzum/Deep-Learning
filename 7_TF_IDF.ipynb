Örnek birkaç cümle üzerinden TF_IDF uygulayarak cümleleri vektörleştirmek.

```python
import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
belgeler=[
    "köpek çok tatlı bir hayvandır",
    "köpek ve kuşlar çok tatlı hayvanlardır",
    "inekler süt üretirler"
]

#tf-idf vektörizer tanımla
tfidf_modeli=TfidfVectorizer()

#belgeleri sayısal vektörlere dönüştür
belge_vektorleri = tfidf_modeli.fit_transform(belgeler)

#kelime kümesini çıkart - vocabulary
kelime_kumesi = tfidf_modeli.get_feature_names_out()

#belgelerin tf idf değerlerini numpy formatına çevir
vektor_temsili=belge_vektorleri.toarray()

print(f"TF-IDF matrisi: \n{vektor_temsili}")
```

```python
#daha okunabilir hale getirmek için pandas df
df_tfidf= pd.DataFrame(vektor_temsili, columns=kelime_kumesi)
print(df_tfidf)
```

```python
# her kelimenin belgeler arasındaki ortalama td idf değerini hesapla
oratalama_tf_idf = df_tfidf.mean(axis=0)
print(f"Kelimelerin ortalama TF-IDF değerleri: \n {oratalama_tf_idf}")
```

```python
import pandas as pd # veri analizi ve veri işleme için kullanılır (dataFrame yapısı sağlar)
from sklearn.feature_extraction.text import CountVectorizer # metinleri sayısal forma dönüştürür (BoW yöntemi)
import re #metinlerde temizleme işlemi yapar
```

```python
import kagglehub
path = kagglehub.dataset_download("lakshmi25npathi/imdb-dataset-of-50k-movie-reviews")
```

```python
veri = pd.read_csv(f"{path}/IMDB Dataset.csv")
veri.head()

yorumlar=veri["review"]
etiketler=veri["sentiment"]

import nltk
from nltk.corpus import stopwords
nltk.download('stopwords')

stop_words=set(stopwords.words("english"))

#metin temizleme
def metin_temizle(metin):
  metin=metin.lower() #küçük harfe çevir
  metin= re.sub(r"\d+","",metin) #rakamları kaldır
  metin=re.sub(r"[^\w\s]","",metin) # özel karakterleri kaldır
  metin=" ".join([kelime for kelime in metin.split() if len(kelime) > 2 and kelime not in stop_words]) #kısa kelimeleri 2 harften kısa olanları kaldır
  return metin

temizlenmis_yorumlar=[metin_temizle(y) for y in yorumlar]
```

```python
#BoW
bow_modeli=CountVectorizer()

yorum_vektorleri=bow_modeli.fit_transform(temizlenmis_yorumlar[:75]) #ilk 75 yorum

kelime_kumesi=bow_modeli.get_feature_names_out() #vocabulary (kelime kümesi)

vektor_temsili=yorum_vektorleri.toarray()
print(f"Vektör temlisi: {vektor_temsili}")
```

```python
df_bow=pd.DataFrame(vektor_temsili, columns=kelime_kumesi)
print(df_bow.head())
```

```python
#kelime frekanslarını hesapla
from collections import Counter # elemanların frekansını sağlar (ör: kelime sıklığı hesaplama)
kelime_sayilari=yorum_vektorleri.sum(axis=0).A1 #her kelimenin toplam kaç adet geçtiğini bul
kelime_frekansi=dict(zip(kelime_kumesi,kelime_sayilari))
en_cok_gecen_5= Counter(kelime_frekansi).most_common(5)

print(f"En çok geçen 5 kelime: {en_cok_gecen_5}")
```

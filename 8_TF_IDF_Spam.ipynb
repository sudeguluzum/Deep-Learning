```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
veri=pd.read_csv("sms_spam.csv")
print(veri.head())
```

```python
#text sütununda yer alan sms mesajlarını al
mesajlar = veri["text"]
```

```python
tf_idf_model = TfidfVectorizer()

#mesajları dönüştür
mesaj_vektorleri= tf_idf_model.fit_transform(mesajlar)

#kelime kümesi - vocabulary
kelime_kumesi= tf_idf_model.get_feature_names_out()

#her kelimenin ortlama tf idf skourunu hesapla
tf_idf_skorlari= mesaj_vektorleri.mean(axis=0).A1

df_tf_idf =pd.DataFrame({"kelime":kelime_kumesi, "ortalama_tf_idf":tf_idf_skorlari})
df_tf_idf_sirali= df_tf_idf.sort_values(by="ortalama_tf_idf", ascending=False)

print(df_tf_idf_sirali.head(10))
```

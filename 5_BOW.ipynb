Metin temsili (bag of words): metin listesi -> sayısal vektörlere

-sklearn CountVectorizer: kelimenin kaç defa geçtiğini sayar ve vektör temsiline dönüştürür

Sonuç:

- kelime kümesi(vocabulary)

- her metin listesi sayısal vektörler ile temsil edilecek

```python
from sklearn.feature_extraction.text import CountVectorizer

#örnek metinlerden oluşan küçük bir veri seti
dokumanlar=[
    "kedi bahçede",
    "kedi evde"
]

#count vectorizer nesnesii oluşturur
kelime_sayac= CountVectorizer()

#dokumanları sayısal vektörlere çevir [bag of words uygulama]
dokuman_vektorleri= kelime_sayac.fit_transform(dokumanlar)

#count vectorizer tarafından bulunan kelime listesi(vocabulary)
kelime_kumesi=kelime_sayac.get_feature_names_out()
print(f"Kelime kümesi: {kelime_kumesi}")

vektor_temsili=dokuman_vektorleri.toarray()
print(f"Vektör temsili: {vektor_temsili}")
```

**Veri Temizleme adımları:**


-fazla boşlukları kaldırma


-büyük harflerin küçük harflere dönüşütürülmesi


-noktalama işsaretlerinin kaldırılması


-yazım hatalarının düzeltilmesi


-html etiketlerinin kaldırılması

```python
#fazla boşlukların temizlenmesi
raw_text= "Python,      Google    NLP      dersi."
print(raw_text.split())
normalized_text_1 = " ".join(raw_text.split())
print("Fazla boşlukları temizle: ", normalized_text_1)

#büyük küçük harf dönüşümü
raw_text="PYTHON, GoogGle NLP"
normalized_text_2= raw_text.lower() #tüm harfleri küçük yapan fonk.
print(f"Temizlenmiş veri: {normalized_text_2}")


#noktalama işaretlerinden kurtul
import string
raw_text="AI Natural-Language-Processing!"
normalized_text_3= raw_text.translate(str.maketrans("","",string.punctuation)) #noktalama işaretlerini boş string ile değiştir
print("Noktalaa işaretleri temizlenmiş: ", normalized_text_3)


#özel karakterlerden kurtul % @ vb.
import re #regular expression (düzenli ifadeler)
raw_text= "Natural @ Language ½ Processing"
normalized_text_4= re.sub(r"[^A-Za-z0-9\s]", "",raw_text)
print(f"Özel karakterlerden kurtul: {normalized_text_4}")


# yazım hatalarından kurtul
from textblob import TextBlob
raw_text="It is amazig 2045"
normalized_text_5= TextBlob(raw_text).correct() #yazım hatalarını düzelt
print("Yazım hatasız: ", normalized_text_5)


#html etikelerinden düz metin elde etme
from bs4 import BeautifulSoup
raw_html="<div> 2026 Google </div>"
normalized_text_6= BeautifulSoup(raw_html, "html.parser").get_text()
print(f"html'den temizlenmiş {normalized_text_6}")
```

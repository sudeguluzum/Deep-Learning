Stop Words çıkarma yöntemleri:


-İngilizce stop words (nltk)

-Türkçe stop words (nltk)

-Manuel olarak stop words çıkarma

```python
import nltk
from nltk.corpus import stopwords

nltk.download("stopwords") #stopwords veri setini indir
```

```python
#İnglizce örnek metin
stop_words_eng = set(stopwords.words("english")) #ing stop words listesi

eng_text="This is just a simple example to show how stop words can be removed from sentences"
eng_text_list= eng_text.split()
print(eng_text_list)

filtered_words_eng= [word for word in eng_text_list if word.lower() not in stop_words_eng]
print(f"Filtered: {filtered_words_eng}")
```

```python
#Türkçe stop word analizi
stop_words_tr= set(stopwords.words("turkish"))

tr_text="Merhaba, bugün sizler ile birlikte Google NLP eğitimi gerçekleştiriyoruz. Bu eğitim sizler için çok faydalı olucaktır."
tr_text_list= tr_text.split()
print(tr_text_list)

filtered_words_tr= [word for word in tr_text_list if word.lower() not in stop_words_tr]
print(f"Filtered: {filtered_words_tr}")
```

```python
#Manuel stop words çıkarma işlemi
custom_tr_stopwords=["bu","ile","de","da","mi"]
custom_text= "Bu bir denemedir, bunun için amacımız metinlerdeki ki bazı kelimeleri çıkartmak."
custom_text_list= custom_text.split()

filtered_custom_words_tr= [word for word in custom_text_list if word.lower() not in custom_tr_stopwords]
print(f"Custom stop words çıkartılmış hali: {filtered_custom_words_tr}")
```
